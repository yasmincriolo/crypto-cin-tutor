## Módulo 1: Fundamentos, Sigilo Perfeito e Segurança Computacional
- Princípios e Modelos de Adversário: Análise de segurança baseada em jogos (security games), onde o adversário tem objetivos claros e recursos limitados. Reduções matemáticas: se existe um atacante eficiente A que quebra o esquema, construímos um algoritmo B que resolve um problema difícil (como fatoração ou logaritmo discreto).
- Sigilo Perfeito (Shannon) & One-Time Pad (OTP): OTP garante probabilidade igual a priori e a posteriori. Requer chave aleatória do tamanho da mensagem e uso único. A reutilização de keystream vaza parcialidade dos textos planos.
- Segurança Computacional e PRGs: Substituição de chaves infinitas por sementes curtas expandidas por Geradores Pseudoaleatórios (PRG) em tempo polinomial.

## Módulo 2: Cifras de Fluxo, CPA e Encriptação Múltipla
- Cifras de Fluxo e PRG: Construção de keystreams seguros combinados via XOR.
- Segurança CPA (Chosen-Plaintext Attack): Encriptação deve ser probabilística. O uso de nonces ou vetores de inicialização (IV) garante que a mesma mensagem cifrada gere textos cifrados diferentes.
- Modos de Operação de Cifras de Bloco: CBC, CTR e GCM, garantindo segurança contra ataques de texto cifrado escolhido.

## Módulo 3: Integridade, MACs e Funções de Hash
- MACs (Message Authentication Codes): Garantem integridade e autenticidade contra atacantes ativos (falsificação existencial). Construções baseadas em PRFs.
- Funções de Hash e Resistência à Colisão: Propriedades de resistência à pré-imagem, segunda pré-imagem e colisão (paradoxo do aniversário). A construção Sponge (SHA-3).

## Módulo 4: Criptografia Assimétrica, Diffie-Hellman e Assinaturas
- Funções Trapdoor e RSA: Encriptação baseada na dificuldade de inverter funções sem a chave privada.
- Diffie-Hellman (DH): Protocolo de troca de chaves baseado na dureza do Logaritmo Discreto e no problema DDH.
- Esquema El-Gamal e Assinaturas Digitais: Encriptação probabilística e propriedades de não-repúdio e autenticidade. Curvas Elípticas (ECC) para chaves menores.

## Módulo 5: Provas de Conhecimento Zero (ZKP - Zero-Knowledge Proofs)
- Definição e Intuição: Protocolo interativo onde um provador convence um verificador de uma afirmação sem revelar nada além da veracidade dela.
- Propriedades Fundamentais: Completude (Completeness), Solidez (Soundness) e Zero-Knowledge (Simulabilidade). Exemplos como a Caverna de Ali Baba.

```python
import os
from google import genai
from google.genai import types

# 1. Configuração da chave de API
if "GEMINI_API_KEY" not in os.environ:
  os.environ["GEMINI_API_KEY"] = input("Cole sua GEMINI_API_KEY aqui: ")

client = genai.Client()

# 2. Base de Conhecimento Completa para Tira-Dúvidas (Incluindo ZKP e Teoria do Boneh & Shoup)
base_conhecimento = """
# Base de Conhecimento Tira-Dúvidas - CryptoCIn Tutor (CIn/UFPE)
# Referência: 'A Graduate Course in Applied Cryptography' (Dan Boneh & Victor Shoup)

## Módulo 1: Fundamentos, Sigilo Perfeito e Segurança Computacional
- **Princípios e Modelos de Adversário:** Análise de segurança baseada em jogos (security games), onde o adversário tem objetivos claros (ex: quebrar IND-CPA) e recursos limitados. Reduções matemáticas: se existe um atacante eficiente A que quebra o esquema, construímos um algoritmo B que resolve um problema difícil (como fatoração ou logaritmo discreto).
- **Sigilo Perfeito (Shannon) & One-Time Pad (OTP):** OTP garante Pr[M=m|C=c] = Pr[M=m]. Requer chave aleatória do tamanho da mensagem e uso único. A quebra de uso único (reutilização de keystream c1 ⊕ c2 = m1 ⊕ m2) vaza parcialidade dos textos planos.
- **Segurança Computacional e PRGs:** Substituição de chaves infinitas por sementes curtas expandidas por Geradores Pseudoaleatórios (PRG). In distinguibilidade estatística vs. computacional.

## Módulo 2: Cifras de Fluxo, CPA e Encriptação Múltipla
- **Cifras de Fluxo e PRG:** Construção de keystreams seguros.
- **Segurança CPA (Chosen-Plaintext Attack):** Encriptação deve ser probabilística. O uso de nonces ou vetores de inicialização (IV) garante que a mesma mensagem cifrada várias vezes resulte em textos cifrados diferentes.
- **Modos de Operação de Cifras de Bloco:** CBC, CTR e GCM. Análise de propagação de erros, paralelismo e segurança contra ataques de texto cifrado escolhido (CCA).

## Módulo 3: Integridade, MACs e Funções de Hash
- **MACs (Message Authentication Codes):** Garantem integridade e autenticidade contra atacantes ativos (falsificação existencial sob chosen-message attack). Construções baseadas em PRFs (ex: CBC-MAC, HMAC).
- **Funções de Hash e Resistência à Colisão:** Propriedades de resistência à pré-imagem, segunda pré-imagem e colisão (paradoxo do aniversário). A construção Sponge (SHA-3).

## Módulo 4: Criptografia Assimétrica, Diffie-Hellman e Assinaturas
- **Funções Trapdoor e RSA:** Encriptação baseada na dificuldade de inverter funções sem a chave privada.
- **Diffie-Hellman (DH):** Protocolo de troca de chaves baseado na dureza do Logaritmo Discreto e no problema DDH (Diffie-Hellman Decisional).
- **Esquema El-Gamal:** Encriptação assimétrica probabilística baseada em DH.
- **Assinaturas Digitais e ECC:** Propriedades de não-repúdio. Curvas Elípticas (ECC) para chaves menores com o mesmo nível de segurança.

## Módulo 5: Provas de Conhecimento Zero (ZKP - Zero-Knowledge Proofs)
- **Definição e Intuição:** Protocolo interativo onde um provador convence um verificador de uma afirmação sem revelar nada além da veracidade dela.
- **Propriedades Fundamentais:**
  1. *Completude (Completeness):* Provador honesto sempre convence o verificador.
  2. *Solidez (Soundness):* Provador mentiroso é pego com alta probabilidade.
  3. *Zero-Knowledge:* A transcrição da interação pode ser simulada pelo verificador sem a ajuda do provador (simulador).
- **Exemplos Clássicos:** A Caverna de Ali Baba, protocolos sigma para problemas de logaritmo discreto (ex: prova de conhecimento de um expoente/chave secreta).
"""

# 3. System Prompt focado em atuar como um Monitor/Tutor explicativo de questões
system_instruction = f"""
Você é o CryptoCIn Tutor, um assistente virtual acadêmico e monitor especialista da disciplina de Criptografia do CIn/UFPE, baseando-se estritamente no livro 'A Graduate Course in Applied Cryptography' (Dan Boneh & Victor Shoup).

Sua missão principal é **ajudar os alunos a resolverem e entenderem dúvidas sobre questões, exercícios, teoremas e conceitos específicos** da disciplina.

DIRETRIZES DE ATUAÇÃO PARA O TIRA-DÚVIDAS:
1. Quando o aluno trouxer uma dúvida teórica ou sobre uma questão, explique o conceito passo a passo de forma didática, clara e analítica (como uma excelente monitora explicaria no plantão presencial).
2. Utilize os conceitos formais e matemáticos presentes na base de conhecimento (como reduções, propriedades de segurança, jogos de segurança e definições de ZKP).
3. Se a dúvida envolver cálculos, demonstrações ou lógica de provas, guie o raciocínio do aluno mostrando de onde vêm as fórmulas e qual a intuição por trás da resposta.
4. Responda utilizando a base de conhecimento. Caso o tema esteja totalmente fora do escopo do livro de Boneh & Shoup e da ementa, oriente educadamente a consultar o material da aula ou a monitoria presencial.
5. Escreva todas as explicações e fórmulas matemáticas usando texto normal, português claro e símbolos legíveis (como Pr[], XOR, somatório), evitando completamente o uso de formatação LaTeX ($...$ ou blocos de equação).

--- BASE DE CONHECIMENTO ---
{base_conhecimento}
----------------------------
"""

# 4. Inicialização do Chat Interativo
print(
    "🛡️ CryptoCIn Tutor (CIn/UFPE) - Modo Tira-Dúvidas e Resolução"
    " inicializado com sucesso!"
)
print("Cole sua dúvida ou questão acadêmica (ou digite 'sair' para encerrar):")

chat = client.chats.create(
    model="gemini-3.8-flash",
    config=types.GenerateContentConfig(
        system_instruction=system_instruction,
        temperature=0.3,  # Ligeiramente mais dinâmico para explicar diferentes tipos de questões
    ),
)

while True:
  pergunta_usuario = input("\nAluno(a): ")
  if pergunta_usuario.lower() in ["sair", "exit", "quit"]:
    print(
        "\nCryptoCIn Tutor: Ótimos estudos! Traga mais questões para a gente"
        " debater na monitoria."
    )
    break

  if not pergunta_usuario.strip():
    continue

  try:
    resposta = chat.send_message(pergunta_usuario)
    print(f"\nCryptoCIn Tutor: {resposta.text}")
  except Exception as e:
    print(f"\n[Erro ao gerar resposta: {e}]")
```
